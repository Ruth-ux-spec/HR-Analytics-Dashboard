```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Attrition by OverTime
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='OverTime', hue='Attrition')
plt.title('Attrition by OverTime')
```

Does working overtime influence employee attrition?

Insight:
From the chart, employees who do not work overtime (No) make up the largest group in the company, and most of them stayed with the organization. However, among employees who work overtime (Yes), the proportion of employees who left the company is noticeably higher. 
This suggests that working overtime is associated with higher employee attrition.

Recommendation:
HR should review overtime policies and workload distribution to reduce excessive overtime. The company should also promote work-life balance and monitor employees who frequently work overtime, as they may be at a higher risk of leaving.

```python
# Attrition rate by Job Role
plt.figure(figsize=(8,5))
role_attr = df.groupby('JobRole')['Attrition'].apply(lambda x: (x=='Yes').mean()*100).sort_values(ascending=False)
sns.barplot(x=role_attr.values, y=role_attr.index)
plt.title('Attrition Rate (%) by Job Role')
plt.xlabel('Attrition Rate (%)')
```

Which job roles have the highest employee attrition rate?

Insight
The chart shows that Sales Representatives have the highest attrition rate among all job roles, followed by Laboratory Technicians, Human Resources, and Sales Executives. These roles appear to experience higher employee turnover than other positions, indicating they may face greater retention challenges.

Recommendation
The HR department should prioritize retention strategies for job roles with the highest attrition rates. This may include reviewing workload, compensation, career development opportunities, employee engagement, and job satisfaction to reduce turnover in these critical roles.

```python
# Attrition by Age Group
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='AgeGroup', hue='Attrition', order=sorted(df['AgeGroup'].unique()))
plt.title('Attrition by Age Group')
```

Which age groups experience the highest employee attrition?

Insight
The chart shows that the youngest age group has the largest number of employees, and it also records the highest number of employees leaving the company. As the age groups increase, the number of employees—and the number of employees leaving—generally decreases. This suggests that employee attrition is more common among younger employees.

Recommendation
HR should strengthen retention initiatives for younger employees by providing structured onboarding, mentoring, career development opportunities, and clear promotion paths. These measures can help improve engagement and reduce early-career turnover.

```python
# Employee Attrition
df["Attrition"].value_counts().plot(kind="bar")
plt.title("Employee Attrition")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")
plt.show()
```

What is the overall employee attrition level?

Insight:
The majority of employees remained with the company, while a smaller proportion left. This indicates that the organization has a relatively good retention rate, although employee turnover still exists.

Recommendation:
HR should continue monitoring employee attrition and focus on identifying the reasons why employees leave in order to improve retention.

```python
# Department bt Attrition
pd.crosstab(
    df["Department"],
    df["Attrition"]
).plot(kind="bar")
plt.title("Department by Attrition")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.show()
```

.Research & Development has the highest number of employees.

.Most employees in every department have not left the company (No).

.Research & Development also has the highest number of employees who left (Yes), but that is partly because it has the largest workforce.

.Sales has the second-highest attrition.

.Human Resources has the fewest employees and the fewest resignations.

Insight:
Research & Development has the largest workforce and also records the highest number of employee exits. Sales follows, while Human Resources has the lowest employee turnover.

Recommendation:
Focus employee retention programs on the Research & Development and Sales departments.
Conduct exit interviews to identify why employees leave these departments.
Improve career development opportunities, employee engagement, and workload management to reduce turnover.

```python
# jobs roles with highest employee turnover
pd.crosstab(df["JobRole"], df["Attrition"]).plot(kind="bar", figsize=(10,5))

plt.title("Attrition by Job Role")
plt.xlabel("Job Role")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45)
plt.show()
```

Insight:
.Sales Executive has the highest number of employees and also records a high number of employees leaving.
    
.Research Scientist and Laboratory Technician also have relatively high employee turnover.

.Research Director and Manager have the lowest attrition.

.Across all job roles, the number of employees who stayed (No) is higher than those who left (Yes).
    
Recommendation:
.HR should prioritize retention strategies for Sales Executives, Research Scientists, and Laboratory Technicians.

.Review workload, career progression, employee satisfaction, and compensation in these roles.

.Continue the current retention practices for job roles with low attrition, such as Managers and Research Directors.

```python
# work life balance influence Attrtion 
pd.crosstab(df["WorkLifeBalance"], df["Attrition"]).plot(kind="bar")
plt.title("Work-Life Balance by Attrition")
plt.xlabel("Work-Life Balance")
plt.ylabel("Number of Employees")
plt.xticks(rotation=0)
```

Explanation
.Most employees have a Work-Life Balance rating of 3.

.Employees with a rating of 3 also have the highest number of employees who stayed (Attrition = No).

.Employees with ratings 1 and 2 show a relatively higher proportion of employees leaving (Attrition = Yes) compared to rating 4.

.Employees with a Work-Life Balance rating of 4 have the lowest attrition, suggesting better work-life balance is associated with employees staying.

Insight:
Work-life balance appears to influence employee retention. Employees with better work-life balance tend to stay with the company, while poorer work-life balance is associated with higher attrition.

Recommendation:
The company should:
Improve work-life balance through flexible work arrangements and manageable workloads.
Monitor departments where employees report poor work-life balance.
Conduct employee satisfaction surveys to identify work-life balance concerns before employees resign.

```python
# marital status with higest Attrtion
pd.crosstab(df["MaritalStatus"], df["Attrition"]).plot(kind="bar")
plt.title("Marital Status by Attrition")
plt.xlabel("Marital Status")
plt.ylabel("Number of Employees")
plt.xticks(rotation=0)

```

Explanation

.Single employees have the highest number of employees who left the company (Attrition = Yes).

Insight:
Single employees appear to have a higher tendency to leave the organization compared to married and divorced employees. This suggests that marital status may have some relationship with employee retention.

Recommendation:

.HR should provide career development opportunities, mentoring, and employee engagement programs, especially for single employees.

.Management should implement retention strategies that improve job satisfaction across all employee groups

```python
# Average Monthly Income by Attrition
df.groupby("Attrition")["MonthlyIncome"].mean().plot(kind="bar")
plt.title("Average Monthly Income by Attrition")
plt.xlabel("Attrition")
plt.ylabel("Average Monthly Income")
plt.xticks(rotation=0)
```

Explanation
.Employees who stayed with the company (Attrition = No) have a higher average monthly income.

.Employees who left the company (Attrition = Yes) have a lower average monthly income.

.This suggests that employees earning lower salaries are more likely to leave the organization.

Insight:
Monthly income appears to influence employee retention. Lower-paid employees have a higher likelihood of leaving, while employees with higher average monthly income are more likely to remain with the company.

Recommendation:
.The company should review its salary structure to ensure employees are paid competitively.

.HR should consider salary adjustments, performance-based incentives, and career progression opportunities to improve employee retention.

.Regular compensation reviews can help reduce employee turnover, especially among lower-income employees.

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
sns.histplot(
    data=df, x='DistanceFromHome', hue='Attrition',
    multiple='stack', bins=12, kde=True,
    palette={'Yes':'#9b8cc4','No':'#8ca6c9'}
)
plt.title('Impact of Distance from Home on Attrition')
plt.xlabel('Distance from Home')
plt.ylabel('Employee Count')
```

Explanation
.The histogram groups employees by their distance from home.

.The colors represent Attrition:

.Yes = Employees who left the company.

.No = Employees who stayed.

The KDE (smooth line) shows the overall distribution of employees in each group.

Insight
.Most employees live 1–3 km from work, and the majority of them stayed with the company.

.Employees who left are present across all distance groups, but they are fewer than those who stayed.

.There is no strong evidence from this chart alone that employees who live farther away are much more likely to leave.

.The highest concentration of employees is at shorter commuting distances.

Recommendation

.Continue supporting employees with flexible work arrangements or transportation assistance, especially those who live farther away.

.Since distance alone does not appear to be the main driver of attrition, HR should also focus on other factors such as:

Job Satisfaction
Work-Life Balance
Overtime
Monthly Income
Career Growth opportunities
